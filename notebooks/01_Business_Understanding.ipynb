# Executive Summary


This project analyzes historical video game sales to identify the factors
associated with commercial success and evaluate whether available game
characteristics can reliably predict global sales.

Using the CRISP-DM framework, the project follows an end-to-end data science
workflow including business understanding, data preparation, exploratory data
analysis, feature engineering, machine learning, model evaluation, and
business recommendations.

Exploratory analysis identified meaningful historical patterns across genres,
platforms, publishers, and geographic markets. These findings demonstrated
that video game performance varies substantially across market segments and
that factors such as genre and platform should be considered together rather
than independently.

The predictive modeling phase evaluated multiple approaches, including linear
and regularized regression, Random Forest, Extra Trees, Gradient Boosting, and
ensemble modeling. Cross-validation and hyperparameter tuning were used to
compare models while maintaining an untouched test set for final evaluation.

Although several models improved upon the baseline, final predictive
performance remained insufficient for reliable game-level sales forecasting.
The results indicate that the available predictors explain only a limited
portion of the variation in global sales.

As a result, deployment of the predictive model is not recommended. The
primary business value of the project comes instead from the historical market
insights identified through exploratory analysis and from demonstrating where
the available data is insufficient for reliable prediction.

The project concludes with data-driven business recommendations, a clear
assessment of modeling limitations, and recommendations for additional data
that could support stronger predictive modeling in future work.

# Project Risk

1. Missing or incomplete historical sales information may affect the reliability
   of the analysis.

2. Extreme blockbuster titles may strongly influence aggregate sales metrics.

3. High-cardinality categorical variables such as Publisher may increase model
   complexity and sparsity.

4. Regional sales variables could introduce target leakage because they
   contribute directly to Global Sales.

5. Historical relationships may not generalize to future video game markets.

6. The available predictors may not contain sufficient information for
   reliable game-level sales forecasting.

7. Increasing model complexity may not improve predictive performance and
   could increase the risk of overfitting.


# Assumptions

1. The dataset provides a reasonably accurate representation of the historical
   video game sales observations included in the analysis.

2. `Global_Sales` and the regional sales variables are measured consistently
   across observations and are expressed in millions of units.

3. Missing or unavailable observations do not systematically distort the major
   historical patterns identified in the analysis.

4. Each row represents an individual video game record associated with its
   listed platform, genre, publisher, year, and sales information.

5. Historical relationships among platform, genre, publisher, year, and sales
   can provide useful descriptive business insights, but they should not be
   assumed to remain unchanged in future markets.

6. Associations identified through exploratory analysis should not be
   interpreted as causal relationships.

7. Regional sales variables are components of `Global_Sales` and therefore
   should not be used as predictors when modeling `Global_Sales`, because doing
   so would introduce target leakage.

8. Model performance on unseen test data provides a more appropriate measure
   of predictive usefulness than training performance alone.

9. If multiple modeling approaches continue to demonstrate weak out-of-sample
   performance, the limitation may reflect insufficient predictive information
   in the available features rather than simply inadequate model complexity.

# Model Evaluation Summary

The objective of the modeling phase was not simply to build a predictive
model, but to determine whether the available historical game attributes
contained enough predictive information to support reliable sales forecasting.

Model performance was evaluated primarily using:

- **MAE (Mean Absolute Error):** Average absolute prediction error.
- **RMSE (Root Mean Squared Error):** Prediction error with greater penalty
  for large errors.
- **R² (Coefficient of Determination):** Proportion of variation in global
  sales explained by the model.

## Interpretation

The machine learning models produced modest improvements over the baseline,
but overall predictive performance remained weak.

Even after testing multiple regression approaches, ensemble methods,
hyperparameter tuning, feature engineering, and target transformation,
the models explained only a small proportion of the variation in global
video game sales.

The results suggest that the primary limitation is the predictive information
available in the dataset rather than simply the choice of algorithm.

The existing models should therefore be treated as exploratory predictive
models rather than systems suitable for reliable sales forecasting or
business deployment.
 


# Business Recommendations

The exploratory and market analyses identified several patterns that can
support business decision-making. However, the machine learning phase showed
that the available features were not sufficient for reliable game-level sales
forecasting. Therefore, the following recommendations are based primarily on
historical market patterns rather than individual sales predictions.

## 1. Match Genre Strategy to Platform

Genre performance varied substantially across platforms.

For example, Shooter games demonstrated strong historical per-title
performance on PS3 and Xbox 360. This suggests that companies should evaluate
genre and platform together rather than assuming that a successful genre will
perform equally well across every platform.

**Recommendation:** Use genre-platform performance as one factor when
evaluating product positioning and historical market opportunities.

## 2. Adapt Strategy to Regional Preferences

North America and Europe displayed relatively similar genre preferences,
while Japan showed a distinctly different market profile.

Role-Playing games represented approximately 27% of Japanese sales compared
with roughly 8% in North America and Europe. Shooter games showed the opposite
pattern, representing a much smaller share of the Japanese market.

**Recommendation:** Product positioning and marketing strategies should be
adapted to regional consumer preferences rather than applying one global
strategy across all markets.

## 3. Evaluate Per-Title Performance Alongside Total Sales

High cumulative sales do not necessarily mean that the typical game in a
category performs strongly.

Some genre-platform and publisher-genre combinations accumulated large sales
because many titles were released, while other combinations achieved stronger
median sales with fewer games.

**Recommendation:** Use both cumulative sales and median sales per title when
evaluating historical market opportunities.

## 4. Consider Publisher Strength and Specialization

Publisher performance also differed across market segments.

Electronic Arts Sports generated the highest cumulative sales among the
publisher-genre combinations analyzed, while Nintendo Platform games achieved
the highest median sales among combinations containing at least 50 games.
Nintendo also demonstrated strength across several genres.

**Recommendation:** Consider publisher experience, brand strength, and
historical specialization when evaluating genre and market opportunities.

## 5. Do Not Deploy the Current Predictive Models

Although multiple regression approaches, feature-engineering strategies,
hyperparameter tuning, and target transformations were evaluated, predictive
performance remained weak.

The models explained only a small proportion of the variation in global
sales, indicating that the available dataset does not contain enough
predictive information for reliable game-level forecasting.

**Recommendation:** The current models should not be used to make high-stakes
sales forecasts or investment decisions.

## 6. Expand the Dataset Before Future Predictive Modeling

Future modeling efforts would likely require additional variables that capture
important drivers of commercial success that are not represented in the
current dataset.

Potential future data could include variables such as marketing activity,
development characteristics, review information, franchise indicators, and
other pre-release market information.

These variables were not available in the current dataset and should therefore
be treated as recommendations for future data collection rather than findings
from the present analysis.

# Final Business Takeaway

The strongest value of this project comes from its market analysis rather
than its predictive model.

Historical sales patterns show that commercial performance is associated with
interactions among platform, genre, publisher, and geographic market.
These relationships can support market analysis and strategic planning, but
the available features are not sufficient to reliably predict the sales of an
individual game.

The appropriate business use of this analysis is therefore to support
strategic decision-making with historical evidence while avoiding unsupported
game-level sales forecasts.

# Feature Engineering Milestone

## What I Learned

1. Preprocessing decisions should be learned from training data rather than test data to prevent data leakage.
2. High-cardinality categorical variables can create large numbers of sparse features after one-hot encoding.
3. Rare publishers can be grouped into an `Other` category to reduce sparsity.
4. Interaction features can capture relationships that individual features may not represent.
5. More features do not necessarily produce a better model.

## Key Decisions

- Reduced Publisher from 528 training categories to 141 categories using rare-category grouping.
- Created a Genre–Platform interaction feature.
- Compared 185 encoded features without the interaction with 476 features including the interaction.
- Created a log-transformed Global Sales target for later model comparison.
