# Feature Engineering Milestone

## What I Learned

1. Preprocessing decisions should be learned from training data rather than test data to prevent data leakage.
2. High-cardinality categorical variables can create large numbers of sparse features after one-hot encoding.
3. Rare publishers can be grouped into an `Other` category to reduce sparsity.
4. Interaction features can capture relationships that individual features may not represent.
5. More features do not necessarily produce a better model.

## Key Decisions

- Reduced Publisher from 528 training categories to 141 categories using rare-category grouping.
- Created a Genre–Platform interaction feature.
- Compared 185 encoded features without the interaction with 476 features including the interaction.
- Created a log-transformed Global Sales target for later model comparison.

## Project Decision Log

| Stage | Decision | Reason |
|---|---|---|
| Planning | Selected Video Game Sales dataset | Suitable for end-to-end analytics and predictive modeling |
| Planning | Adopted CRISP-DM workflow | Provides a structured approach from business understanding through deployment |
| EDA | Retained legitimate sales outliers | Blockbuster games represent meaningful real-world observations |
| EDA | Used median for categorical comparisons | Reduces the influence of extreme sales values |
| Feature Engineering | Excluded regional sales as predictors | Prevents target leakage because regional sales contribute directly to Global Sales |
| Feature Engineering | Grouped rare publishers into `Other` | Reduces high-cardinality and sparse one-hot encoded features |
| Feature Engineering | Learned rare categories from training data only | Prevents test-set information from influencing preprocessing |
| Feature Engineering | Created Genre–Platform interaction | EDA suggested sales performance differs across genre-platform combinations |
| Feature Engineering | Retained base and interaction feature sets | Allows model performance to determine whether added complexity is justified |
| Feature Engineering | Created log-transformed target | Reduces the influence of extreme right-skew while retaining blockbuster games |
